```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation


# 모터사이클, 야구, 그래픽스, 윈도우즈, 중동, 기독교, 의학, 우주 주제를 추출.
cats = ['rec.motorcycles', 'rec.sport.baseball', 'comp.graphics', 'comp.windows.x',
        'talk.politics.mideast', 'soc.religion.christian', 'sci.electronics', 'sci.med'  ]
# 8개 카테고리

# 위에서 cats 변수로 기재된 category만 추출. featch_20newsgroups( )의 categories에 cats 입력
news_df= fetch_20newsgroups(subset='all',remove=('headers', 'footers', 'quotes'),
                            categories=cats, random_state=0)
                            # 8개 카테고리 데이터(뉴스)만 가져와라.

#LDA는 Count 기반의 Vectorizer만 적용합니다.  
count_vect = CountVectorizer(max_df=0.95, max_features=1000, min_df=2, stop_words='english', ngram_range=(1,2))
                            # 많이 나오는 단어 5% 빼라         조금 나오는 거 빼라..
feat_vect = count_vect.fit_transform(news_df.data)
print('CountVectorizer Shape:', feat_vect.shape)
# CounterVectorizer Shape: (18846, 1000)
# CounterVectorizer Shape: (7862, 1000)
```

```python
# ['alt.atheism', 'comp.graphics', 'comp.os.ms-windows.misc', 'comp.sys.ibm.pc.hardware', 'comp.sys.mac.hardware',
#  'comp.windows.x', 'misc.forsale', 'rec.autos', 'rec.motorcycles', 'rec.sport.baseball', 'rec.sport.hockey',
#  'sci.crypt', 'sci.electronics', 'sci.med', 'sci.space', 'soc.religion.christian', 'talk.politics.guns',
#  'talk.politics.mideast', 'talk.politics.misc', 'talk.religion.misc']
```

```python
lda = LatentDirichletAllocation(n_components=8, random_state=0)
                                # 현실은 8개인지 모름. 비지도는 답 없음. 
                                # 책에서는 답을 알고 있으니 8로 했지만 실무에서는 모름.
lda.fit(feat_vect)
```

```python
print(lda.components_.shape)
lda.components_
# 8개 축으로 나눔.(잠재적인 의미의 축)
# 컬럼은 단어1, 단어2, ... , 단어1000
# 축마다 축에 대한 단어들의 가중치가 들어있음.
```

```python
def display_topics(model, feature_names, no_top_words):
    for topic_index, topic in enumerate(model.components_):     # 8번 반복, 1번에 1000개의 단어 가중치 값이 들어감.
        print('Topic #',topic_index)


        # components_ array에서 가장 값이 큰 순으로 정렬했을 때, 그 값의 array index를 반환.
        topic_word_indexes = topic.argsort()[::-1]
        top_indexes=topic_word_indexes[:no_top_words]   # 상위 추출
       
        # top_indexes대상인 index별로 feature_names에 해당하는 word feature 추출 후 join으로 concat
        feature_concat = ' '.join([feature_names[i] for i in top_indexes])                
        print(feature_concat)


# CountVectorizer객체내의 전체 word들의 명칭을 get_features_names( )를 통해 추출
feature_names = count_vect.get_feature_names_out()


# Topic별 가장 연관도가 높은 word를 15개만 추출
display_topics(lda, feature_names, 15)

# 텍스트 숫자화. 텍스트 특징 추출.
```
