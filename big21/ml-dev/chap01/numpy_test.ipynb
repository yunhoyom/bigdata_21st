```python
org_array = np.array([ 3, 1, 9, 5])
print('원본 행렬:', org_array)
# np.sort( )로 정렬
sort_array1 = np.sort(org_array)    # ascending 밖에 못함.(메소드 보니까 asc 설정이 없음.) 내림차순 불가
print ('np.sort( ) 호출 후 반환된 정렬 행렬:', sort_array1)
print('np.sort( ) 호출 후 원본 행렬:', org_array)
# ndarray.sort( )로 정렬
sort_array2 = org_array.sort()  # 반환되는 값이 없음.  # ascending 밖에 못함.(메소드 보니까 asc 설정이 없음.) 내림차순 불가
print('org_array.sort( ) 호출 후 반환된 행렬:', sort_array2)
print('org_array.sort( ) 호출 후 원본 행렬:', org_array)
```

```python
# 내림차순(descending) : sort 한 다음, 슬라이싱으로 처리하면 내림차순 가능
org_array = np.array([3,1,9,5])
sort_arr1 = np.sort(org_array)
print(f'원본 데이터 : {org_array}') # 원본 유지된다.
print(f'정렬한 데이터 : {sort_arr1}')   # 전체 데이터
print(f'정렬한 데이터 : {sort_arr1[:]}')    # 슬라이싱 전체 데이터
print(f'정렬한 데이터 내림차순 : {sort_arr1[::-1]}')    # 슬라이싱한 상태에서 step : -1 거꾸로 가져옴.
```

```python
# 2차원 데이터 정렬
# 2차원(행, 열) 축 존재

array2d = np.array([[8, 12],
                   [7, 1 ]])


sort_array2d_axis0 = np.sort(array2d, axis=0)      # axis=0 : 행 방향
print('로우 방향으로 정렬:\n', sort_array2d_axis0)  # 행 방향(위에서 아래 방향)으로 정렬. (7 vs 8), (12 vs 1)


sort_array2d_axis1 = np.sort(array2d, axis=1)      # axis=1 : 열 방향
print('컬럼 방향으로 정렬:\n', sort_array2d_axis1)  # 열 방향(좌에서 우 방향)으로 정렬. (8 vs 12), (7 vs 1)
```

```python
# 정렬? : 크기 비교, 오름차순한 다음 첫번째 값(min), 내림차순한 다음 첫번째 값(max)
# 정렬된 인덱스 반환 함수 : np.argsort(정렬할 data)
# 정렬한 다음 인덱스만 추출할 수 있음.
org_array = np.array([ 3, 1, 9, 5])
sort_indices = np.argsort(org_array)    # argsort: 정렬된 배열의 인덱스 추출.
print(type(sort_indices))
print('행렬 정렬 시 원본 행렬의 인덱스:', sort_indices)
# 1: 가장 작은 수, 2: 가장 큰 수.(인덱스로 서열에 따른 숫자 파악 가능)
```

```python
org_array = np.array([ 3, 1, 9, 5])
sort_indices_desc = np.argsort(org_array)[::-1] # 내림차순 핵심.
print('행렬 내림차순 정렬 시 원본 행렬의 인덱스:', sort_indices_desc)
```

```python
# argsort 예제
# 점수에 의해 정렬된 이름을 구한다.
import numpy as np


score_array= np.array([78, 95, 84, 98, 88])
name_array = np.array(['John', 'Mike', 'Sarah', 'Kate', 'Samuel'])  # 가장 높은 점수의 인덱스 가져오기


sort_indices_asc = np.argsort(score_array)[::-1]    # 현재 내림차순.
print('성적 오름차순 정렬 시 score_array의 인덱스:', sort_indices_asc)
print('성적 오름차순으로 name_array의 이름 출력:', name_array[sort_indices_asc])

```

```python
# 내적: np.dot(), @
# 데이터 
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])


dot_product = np.dot(A, B)
print('행렬 내적 결과:\n', dot_product)
```
