```python
# gender 처리, customers.csv
cust = pd.read_csv("../data/customers.csv")
```

```python
cust["gender"].value_counts(dropna=False)
```

```python
cust["gender"][:10]
```

```python
# 대체하는 딕셔너리 만들기
gmap = {"남":"M", "여":"F", "M":"M", "F":"F"}
cust["gender"] = cust["gender"].map(gmap)   # map은 Series 메소드
cust["gender"].value_counts(dropna=False)
```

```python
# "프리미엄" 상품 수
# 상품명에 "프리미엄" 문자열 포함.
prod[prod["product_name"].str.contains("프리미엄")]
prod["product_name"].str.contains("프리미엄").sum()
```

```python
prod.product_name[:20]
```

```python
# 상품명을 등급과 상품명으로 분리
# 1. 양쪽 공백 제거, split(" ") 주변 공백 없애기 위해.
prod["product_name"] = prod["product_name"].str.strip()
# 2. split() 처리
parts = prod["product_name"].str.split(" ", n=1, expand=True)    # n=1: 1개만 split해라. 예. 플러스 홍차 커피 -> ["플러스", "홍차 커피"]
prod["grade"] = parts[0]    # '0' 컬럼명. grade는 새로 만드는 것.
prod["item"] = parts[1]
prod.head()
```

```python
prod["grade"].value_counts()
```
