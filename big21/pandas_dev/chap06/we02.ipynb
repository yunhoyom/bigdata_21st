```python
import pandas as pd
```

```python
# 데이터 로딩 = 데이터 프레임 생성
cust = pd.read_csv("../data/customers.csv")
```

```python
# 데이터 탐색
cust.info()
```

```python
# 도시명에 공백 문제 발생.
# 문제 확인
print(f"정제 전 도시 종류 수 : {cust['city'].nunique()}")
print(f"도시명: {cust['city'].unique()}")
```

```python
# print(f"도시명: {cust['city'].value_counts()}")       # Series
print(f"도시명: {cust['city'].value_counts().index}")   # Series에서 index 추출
print(f"도시명: {cust['city'].value_counts().values}")  # Series에서 values 추출
```

```python
# 정제 후 확인.
# cust["city"].str.strip()    # 문자열 함수 : strip(), 양쪽 공백 제거 가능. 자체만으로는 원본 반영 안 됨.
cust["city"] = cust["city"].str.strip()     # 원본 반영 안 되기 때문에 대입해야 함.
```

```python
print(f"정제 후 도시 종류 수 : {cust['city'].nunique()}")
print(f"정제 후 도시 : \n {cust['city'].value_counts()}")
```
