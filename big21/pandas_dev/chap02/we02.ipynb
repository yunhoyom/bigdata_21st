```python
import pandas as pd

# 옵션 없이 읽기
orders_raw_df = pd.read_csv("../data/orders.csv")
# orders_raw_df.head(3)
# print(orders_raw_df.columns)
print(orders_raw_df.dtypes)
```

```python
# order_datetime : str => datetime object
orders_df = pd.read_csv(
    "../data/orders.csv",
    parse_dates=['order_datetime']
)
print(orders_df.dtypes)
```

```python
# 원하는 컬럼만 추출 : 옵션 => usecols=[] 여러 개 선택 가능
orders_df = pd.read_csv(
    "../data/orders.csv",
    usecols = ["order_id","channel","status"],    # 3개 컬럼 추출
    nrows = 5    # 미리보기. 빠르고 메모리도 덜 씀. 근데 잘 안 씀..
)
orders_df
```
