```python
# 결측치(누락되어 있는 값) 처리
# 결측 개수
# 고객 데이터
customers_df = pd.read_csv("../data/customers.csv")
customers_df.head()
```

```python
customers_df.info()
```

```python
# 생일 데이터 없는 고객의 수
# isna() : Null이니? Null이면 True, 아니면 False(있으면 True, 없으면 False)
customers_df["birth_date"].isna()
```

```python
# 단일 컬럼의 결측 개수
customers_df["birth_date"].isna().sum()    # True: 1, False : 0
```

```python
# 전체 컬럼 각각의 결측 개수 -> Series
customers_df.isna().sum()[["email","gender"]]
```
