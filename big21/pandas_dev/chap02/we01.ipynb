```python
# DataFrame을 dict로 직접 생성
import pandas as pd

sample = {
    "product_name":["코어 티셔츠","베이직 노트북","데일리 원두커피"],    # key,value 한 쌍이 Series가 됨. key가 컬럼명, vaule가 값
    "category":["의류","전자","식품"],
    "price":[39000,89000,15000],
}     # Series 모여 DataFrame 됨.

# print(type(sample))
# sample_df = pd.DataFrame()    # 빈 DataFrame 생성
sample_df = pd.DataFrame(sample)    # 빈 DataFrame 생성
sample_df.head()
```

```python
# 인덱스 추출
# sample_df.index ; ndarray(다차원배열, numpy) => list로 변환.
# Index : RangeIndex(start=0, stop=3, step=1)
# tolist() : 리스트로 변환
print(f"Index : {sample_df.index.tolist()}")
# Index : [0, 1, 2]
```

```python
# 컬럼 정보 추출
# sample_df.columns 변수 => 레퍼런스 => 오브젝트
sample_df.columns.tolist()
```

```python
# 데이터프레임(sample_df)의 데이터 타입 추출
sample_df.dtypes     # column마다 dtype
type(sample_df.dtypes)
sample_df.dtypes.index
sample_df.dtypes.values
```
