#### 데이터 형태

##### 범주형(문자열, 숫자)
- 예. 도시, 성별, 학과, 요일, 혈액형 등.
- 명목형: 순서x
- 순서형: 순서o

##### 연속형(숫자: 정수, 실수)
- 예. 키, 몸무게, 매출액, 점수 등

##### 구분법: 값 두 개 고른 뒤 평균을 낸다. 말이 되면 연속형, 안 되면 범주형

```python
log2 = logs.copy()  # 복사본 대입, 원본과 분리
log2["event_type"] = log2["event_type"].astype("category")

# 메모리 사용량 : logs["event_type"] str, log2["event_type"] category
before = logs.memory_usage(deep=True)["event_type"]
after = log2.memory_usage(deep=True)["event_type"]
print(f"logs event_type memory usage: {before}")
print(f"log2 event_type memory usage: {after}")
```
