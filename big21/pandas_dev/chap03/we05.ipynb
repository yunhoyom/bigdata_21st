```python
import pandas as pd
```

```python
customers_df = pd.read_csv("../data/customers.csv")
products_df = pd.read_csv("../data/products.csv")
```

```python
# 특정 고객 번호(5000~5010) 범위의 이름, 도시 추출
sel = customers_df.loc[
    customers_df["customer_id"].between(5000, 5010),
    ["customer_id","name", "city"]
].sort_values("customer_id")
sel
```

```python
# 서울, 부산에 있는 고객 추출

customers_df.loc[
    customers_df["city"].isin(["서울", "부산"]),
    ["customer_id", "name", "city"]
]
```

```python
# 서울, 부산에 있는 고객 외 추출
customers_df.loc[
    ~customers_df["city"].isin(["서울", "부산"]),
    ["customer_id", "name", "city"]
]
# 위 아래 동일
customers_df[~customers_df['city'].isin(["서울", "부산"])][["name", "city"]]
```

```python
# 홍차가 포함된 상품명 조회
# products_df["product_name"].str.contains("홍차")
products_df.loc[
    products_df["product_name"].str.contains("홍차"),
    ["product_name","category"]
]
```

```python
# 성별이 '여'이고 서울 거주하는 고객 추출
condition = (customers_df["gender"] == "여") & (customers_df["city"] == "서울")
# ((customers_df["gender"] == "여") & (customers_df["city"] == "서울")).sum()
customers_df.loc[
    condition,
    ["name","gender","city"]
]
```

```python
# 성별 "여"이고 "서울" 거주하는 고객 추출 (잘 씀.)
customers_df.query("gender=='여' and city=='서울'")[["name", "gender","city"]]
```

```python
# 원가 상위 5개 추출
# products_df.sort_values("cost", ascending=False).head(5)
products_df.nlargest(5, "cost")

# 원가 하위 5개 추출
products_df.nsmallest(5, "cost")
```
