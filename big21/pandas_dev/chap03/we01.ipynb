```python
# 주문 데이터 읽기
orders_df = pd.read_csv(
    "../data/orders.csv",           # order_datetime  str(datetime 아니라 str임)
    parse_dates=["order_datetime"]  # order_datetime의 dtype을 datetime64로 변경
)
orders_df.info()
```

```python
# 행 기준 추출 : loc[행,열] -> 문자열, iloc[행,열] -> 숫자
orders_df.loc[0:5]  # ':' => 범위 지정(행이 0이라는 값부터~5라는 값까지. 열은 없으니 모든 행.)
# orders_df.iloc[0:5]   # ':' => 범위 지정(행이 0~4행까지. start ~ end -1)
```

```python
# 행, 열 접근
print(orders_df.head())
print(orders_df.iloc[0,1])
print(orders_df.loc[0,"customer_id"])
```

```python
# 행, 열 접근
# 고객 정보
customer_df = pd.read_csv("../data/customers.csv")
print(customer_df.head())
print(customer_df.iloc[0,1])
```

```python
# 앞에서 3행, 이름과 도시 추출
# customer_df.iloc[:3, [1,5]]           # 숫자로 추출(iloc)
customer_df.loc[:2, ["name","city"]]    # 숫자와 문자열 추출(loc)
```
