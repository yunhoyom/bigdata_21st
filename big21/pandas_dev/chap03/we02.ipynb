```python
import pandas as pd

products_df = pd.read_csv("../data/products.csv")
```

```python
products_df.head()
```

```python
# 원가가 높은 상품의 이름과 카테고리 추출, cost > 100000 이상
products_df["cost"] > 100000    # Boolean (True, False) return Series
```

```python
expensive = products_df.loc[
    products_df['cost'] > 100000,           # 행 자리
    ["product_name", "category", "cost"]    # 열 자리, 다중 열 선택 => 데이터프레임 추출
]
expensive
```
