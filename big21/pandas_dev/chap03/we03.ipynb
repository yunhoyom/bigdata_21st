```python
import pandas as pd

customers_df = pd.read_csv("../data/customers.csv")
```

```python
# 고객번호 중복 확인
# customers_df["customer_id"].duplicated()      # 중복 여부(True / False)
customers_df["customer_id"].duplicated().sum()  # True & False 는 sum()이 됨. True의 개수
```

```python
# 중복된 고객 정보 추출
dup_cust = customers_df.loc[
    customers_df["customer_id"].duplicated(),   # 중복된 행만 선택
    ["customer_id", "name"]
]
dup_cust
```
