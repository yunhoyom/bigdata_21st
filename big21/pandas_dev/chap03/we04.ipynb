```python
import pandas as pd
# 원가 상위 5개 추출. 상품명, 카테고리, 원가

products_df = pd.read_csv("../data/products.csv")
top5_products = products_df.sort_values("cost", ascending=False).head(5)
top5_products
```

```python
top5_products[["product_name","category","cost"]]
```

```python
# 원가 하위 5개 추출. 상품명, 카테고리, 원가
bottom5_products = products_df.sort_values("cost", ascending=True).head(5)
bottom5_products[["product_name", "category", "cost"]]
```

```python
# 여러 컬럼으로 정렬, 정렬 방식도 다양하게 적용
multi_df = products_df.sort_values(
    ["category", "cost"], # 정렬할 컬럼을 지정. 1차 정렬: category, 2차 정렬: cost
    ascending=[True, False]
)
multi_df.head()
```

```python
# 날짜 컬럼을 인덱스로 처리해서 특정 날짜 단위로 슬라이싱(범위 지정) 처리
# 주문 데이터로 처리
orders_df = pd.read_csv(
    "../data/orders.csv",
    parse_dates = ["order_datetime"]
)
orders_df.info()        # order_datetime  198067 non-null 2000개는 null
```

```python
# 주문 날짜 데이터 null인 행 삭제: 날짜 column을 index로. 날짜로 행 선택하려고 보냄.
orders_drop_df = orders_df.dropna(subset=["order_datetime"])    # dropna: 결측값인 행 삭제. subset은 리스트로 받으니 여러 값을 받을 수 있음.
orders_drop_df.info()
```

```python
orders_drop_df.head()
```

```python
# order_datetime을 index로 보낸다.
orders_drop_dt_df = orders_drop_df.set_index("order_datetime")
orders_drop_dt_df.head()
```

```python
orders_drop_dt_sort_df = orders_drop_dt_df.sort_index()
orders_drop_dt_sort_df.head()
```

```python
ord_idx = (orders_df.dropna(subset=["order_datetime"])   # DataFrame    # 결측치 제거
            .set_index("order_datetime")                 # DataFrame    # index
            .sort_index())                               # DataFrame    # sorting
ord_idx
```

```python
# 2024-06월 주문 데이터 건수
june_df = ord_idx.loc["2024-06"]    # datetime이라 가능.
june_df.shape[0]
```

```python
# 2024-06월 결측치까지 있는 주문 데이터 건수
tot_cnt = orders_df.set_index("order_datetime").sort_index().loc['2024-06'].shape[0]
tot_cnt
```
