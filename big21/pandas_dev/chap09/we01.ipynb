```python
import pandas as pd
```

```python
# 주문 상세(order_item)에 주문(orders)과 고객(customers)을 붙이기
# 데이터 로딩
customers = pd.read_csv("../data/customers.csv")
orders = pd.read_csv("../data/orders.csv")
items = pd.read_csv("../data/order_items.csv")
```

```python
# items에 orders를 병합.
items.columns, orders.columns
```

```python
# 1. 주문 상세(50만 건) + 주문(20만 건) : order_id
oi = items.merge(
    right=orders,
    on="order_id",
    how="inner"
)
items.shape, oi.shape
```

```python
# + 고객
m = oi.merge(
    right=customers,
    on="customer_id",
    how="inner"
)
m.shape

# merge 전에 customer_id 중복 제거해야. (501199, 16)
```
