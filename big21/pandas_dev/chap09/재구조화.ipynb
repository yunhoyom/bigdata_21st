```python
import pandas as pd
```

```python
# pivot
df =pd.DataFrame({
    "date":["2026-01","2026-01","2026-02","2026-02"],
    "city":["seoul","busan","seoul","busan"],
    "temp":[-2.5,1.0,0.5,3.5],
    "humid":[40,55,45,60]
})
df
```

```python
# 
long = df.groupby(["date","city"])[["temp","humid"]].sum()
long

df.pivot(
    columns="city",
    index="date",
    values="temp"
)
```

```python
# values = multi
result = df.pivot(
    columns="city",
    index="date",
    values=["temp", "humid"]
)
result["temp"]
```

```python
result["humid"]
```

```python
result["temp", "busan"]["2026-02"]
```

```python
# melt
df = pd.DataFrame({
    "name":["hong","ko"],
    "kor":[90,80],
    "eng":[85,95],
    "grade":["A","B"]
})
df
```

```python
df.melt(
    id_vars="name",
    value_vars=["kor","eng"],    # values 에 들어간 컬럼 지정
    var_name="subject"
)
```

```python
df.melt(
    id_vars=["name", "grade"],
    value_vars=["kor", "eng"],
    value_name="score",
    var_name="subject"
)
```

```python
# stack : 컬럼을 인덱스로 넘겨줌
# 메소드 원형 : DataFrame.stack(lavel=-1, dropna=<no_default>, sort=True, future_stack-True)
# level 인덱스로 내릴 컬럼의 레벨. 정수, 레벨명(문자열), 또는 리스트, -1: 가장 안쪽(마지막) 레벨
```

```python
stack_df = pd.DataFrame({
    "kor":[90,80],
    "eng":[85,85]
})
stack_df
```

```python
stack_df.columns.name="subject"
```

```python
print(type(stack_df))             # stack X
print(stack_df.stack())     # stack O, (kor,eng) => index로 보냈다. column을 index로 보낸다.
print(type(stack_df.stack()))
```

```python
import numpy as np
```

```python
# np.arange(8)      # ndarray. list가 아님.
# np.arange(8).shape  # 1차원  (8,)
np.arange(8).reshape(2,4) # 2X4. 2차원
```

```python
# 다중 컬럼: 특정 레벨
mi = pd.DataFrame(
    data=np.arange(8).reshape(2,4),
    index=["r1","r2"],
    columns=pd.MultiIndex.from_product(
        [["A","B"],["x","y"]],
        names=["grp","sub"]
    )
)
mi
```

```python
# stack(MultiLevel에서). sub(column)를 index로 보냄.
mi.stack(level="sub")
```

```python
# 모든 columns를 index로 보냄.
mi.stack(level=["grp","sub"])
```
