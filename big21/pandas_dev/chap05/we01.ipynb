```python
import pandas as pd
```

```python
# 고객 데이터 로딩(적재)
cust_df = pd.read_csv("../data/customers.csv")
```

```python
cust_df.info()
```

```python
# 1. 열별 결측 개수와 비율 진단
# cust_df : DataFrame.isna() => 컬럼 7개, 컬럼별 isna().sum()
print(f"결측 개수 : {cust_df.isna().sum()}")
print(f"\n결측 비율: {cust_df.isna().mean() * 100}")
```

```python
# 결측치 처리
# 원본을 보호
cleaned = cust_df.copy()    # 복사해서 대입, 원본 보호

# 1. gender column : "미상"으로 채우기. => fillna("미상")
cleaned["gender"] = cleaned["gender"].fillna("미상")

# 2. birth_date 값이 없는 행은 삭제: dropna() => 행 삭제
# age_ready = cust_df.dropna(subset=["birth_date"],)   # 삭제된 데이터프레임 반환
age_ready = cust_df.dropna(subset=["birth_date"])   # inplace=True 데이터프레임(cust_df)에 반환(웬만하면 쓰지 말기.)

# 3. email: 분석에 사용하지 않으니 그냥 두기.
print(f"gender 미상 채운 뒤 결측 : {cleaned['gender'].isna().sum()}")
print(f"birth_date 기준 dropna 전 행 수 : {len(cust_df)}")  # 삭제 전
print(f"birth_date 기준 dropna 후 행 수 : {len(age_ready)}")    # 삭제 후
```

```python
# 숫자 컬럼 결측은 통계값으로 채우는 방법도 있다.
# 주문 상세에서 단가 컬럼에 결측 처리 => 평균 단가로 채우기.
# 주문 상세 데이터 로딩
order_items_df = pd.read_csv("../data/order_items.csv")
```

```python
order_items_df.info()
```

```python
# 단가(unit_price) 평균, 중앙값
print(f"평균: {order_items_df['unit_price'].mean()}")
print(f"중앙값: {order_items_df['unit_price'].median()}")
```

```python
# 단가 결측이 있는 행의 수
print(f"unit_price 결측: {order_items_df['unit_price'].isna().sum()}")
```

```python
# fillna(value): 결측에 value 채워라
order_items_df["unit_price"] = order_items_df["unit_price"].fillna(order_items_df["unit_price"].mean())
order_items_df["unit_price"].isna().sum()
# 이상치(outlier) 없는 경우 평균으로 채워도 된다.
```
