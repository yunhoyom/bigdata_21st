```python
import pandas as pd
```

```python
# 데이터 로딩: 주문 상세
oi = pd.read_csv("../data/order_items.csv")
oi.info()
```

```python
oi.describe()[["quantity","unit_price","discount"]]
```

```python
# 1. 진단: 중복, 이상값
before_rows = len(oi)
print(f"정제 전 행 수 {before_rows}")
# oi.duplicated() : boolean(True or False)
dup_cnt = oi.duplicated().sum()
print(f"완전 중복 행 수: {dup_cnt}")
neg_qty = (oi["quantity"] <= 0).sum()
print(f"수량 0 이하(이상값) 개수: {neg_qty}")
print(f"수량 최소값/최대값 : {oi['quantity'].min()}, {oi['quantity'].max()}")
```

```python
# 2. 규칙 적용: 중복 제거 후 수량 1개 이상만 남긴다.
oi_clean = oi.drop_duplicates() # 중복 삭제
print(f"중복 제거 후 길이: {len(oi_clean)}")
oi_clean = oi_clean[oi_clean["quantity"] > 0]
print(f"정제 후 행 수 : {len(oi_clean)}")
```

```python
# 1. 데이터 타입 변환 : 숫자 변환 실패(문자열 오염) 개수 진단:
# 변환할 수 있는 price는 변환하고 에러나는 price는 다른 방식으로 처리.
# errors => 
#   raise(디폴트): 변환 실패 시 예외 발생 -> 프로그램 중단.
#   coerce: 변환 실패 시 실패 값을 NaN으로 대체 -> 프로그램 중단 x
#   ignore: 원본 그대로 반환.
fail_cnt = pd.to_numeric(prod["price"], errors="coerce").isna().sum()   #np.int64(46)
```

```python
# 변환 실패한 값 확인: 무엇이 문제인지 진단
prod.loc[
    pd.to_numeric(prod["price"], errors="coerce").isna(), # 행
    "price"  # 열
].head(10)
```

```python
# 규칙 : '원', ',' 제거 후 숫자화
price_str = (prod["price"].astype(str)
             .str.replace("원","", regex=False)
             .str.replace(",","", regex=False))
prod["price"] = pd.to_numeric(price_str, errors='coerce')   # float
prod["price"].isna().sum()
```

```python
# 음수 가격 확인
(prod["price"] < 0).sum()
```
