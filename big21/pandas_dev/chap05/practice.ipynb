```python
import pandas as pd
```

```python
cust = pd.read_csv("../data/customers.csv")
print(f"정제 전 행수: {len(cust)}")
print(f"완전 중복 행수: {cust.duplicated().sum()}")
print(f"customer_id 기준 중복: {cust.duplicated(subset=['customer_id']).sum()}")
```

```python
cust_clean = cust.drop_duplicates()
print(f"정제 후 행수 : {len(cust_clean)}")
```
