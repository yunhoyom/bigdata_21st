```python
import pandas as pd
```

```python
# 고객의 연령대, 성별 분포(명수) 확인
# 데이터 로딩
cust = pd.read_csv("../data/customers.csv").drop_duplicates()
cust.info()
```

```python
# 데이터 전처리
# 1. NaN 제거
cust = cust.dropna(subset=["gender","birth_date"])
cust.info()
```

```python
# 2. birth_date 컬럼의 데이터타입이 현재 str -> datetime64 변경

cust["birth_date"] = pd.to_datetime(cust["birth_date"])
cust.info()
```

```python
# 나이(age) 구하기
cust["age"] = 2024 - cust["birth_date"].dt.year     # dt: 날짜를 다 쪼갬.
cust[["birth_date","age"]].head()
```

```python
# 연령대: "20대 이하","30-40대","50대이상"
bins=[0,29,49,200]
labels=["20대 이하","30-40대","50대이상"]
cust["age_group"] = pd.cut(
    x=cust["age"],
    bins=bins,
    labels=labels
)
cust.head()
```

```python
# 성별을 한국어로 통일
cust["gender_ko"] = cust["gender"].map(
    {"M":"남", "남":"남", "F":"여", "여":"여"}
)
cust.head()
```

```python
# 고객의 연령대, 성별 분포(명수) 확인
freq = pd.crosstab(cust["age_group"], cust["gender_ko"])
freq
```

```python
# 고객의 연령대, 성별 분포(비율) 확인
pd.crosstab(
    cust["age_group"],
    cust["gender_ko"],
    normalize="index"        # 각 행의 합을 1로 보고 비율
).round(3)
```

```python
# 고객의 연령대, 성별 분포(비율) 확인
pd.crosstab(
    cust["age_group"],
    cust["gender_ko"],
    normalize="columns"        # 각 열의 합을 1로 보고 비율
).round(3)
```
