```python
import pandas as pd
```

```python
# 웹으로 주문된 주문 중 배송 중인 주문 개수
orders_df = pd.read_csv("../data/orders.csv", parse_dates=["order_datetime"])
orders_df.head()
```

```python
con1 = orders_df["channel"] == "web"
con2 = orders_df["status"] == "delivered"

orders_df[con1 & con2].shape[0]
```

```python
# 웹 외의 주문 수
con = ~(orders_df["channel"] == "web")
orders_df[con].shape[0]
```

```python
# 'app', 'canceled' 주문 수

con1 = orders_df["status"] == "canceled"
con2 = orders_df["channel"] == "app"
orders_df[con1 & con2].shape[0]
```
