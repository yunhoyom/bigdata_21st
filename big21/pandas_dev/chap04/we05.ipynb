```python
import pandas as pd
```

```python
# 0. orders.csv 데이터 로딩
orders_df = pd.read_csv("../data/orders.csv", parse_dates=["order_datetime"])
```

```python
# 1. 앱 주문, 6월 주문
con1 = orders_df["channel"] == 'app'
con2 = orders_df["order_datetime"].dt.month == 6
seg_a = orders_df[con1 & con2]
seg_a.head()
```

```python
# 2. 매장에서 주문, 취소된 주문
con1 = orders_df["channel"] == "web"
con2 = orders_df["status"] == "canceled"
seg_b = orders_df[con1 & con2]
seg_b.info()
```

```python
# 3. 취소, 반품도 아닌 주문
# ~orders_df["status"].isin(["canceled", "returned"])
seg_c = orders_df[~orders_df["status"].isin(["canceled", "returned"])]
seg_c["status"].unique()
```
