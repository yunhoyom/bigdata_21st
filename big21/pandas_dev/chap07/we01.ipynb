```python
import pandas as pd
```

```python
# 상품 데이터프레임에 마진(가격-원가) 컬럼 추가: assign
# 상품 데이터 로딩 
prod = pd.read_csv("../data/products.csv")
```

```python
# 가격 데이터 문제점(str) : '원', ',' 없애고 숫자로 변경해야 계산 가능
# astype(str) : str 데이터 변경
# .str : str 메소드를 사용 가능
prod["price"] = pd.to_numeric(
    (prod["price"].astype(str)
     .str.replace("원","", regex=False)     # 결과가 Series : prod["price"]
     .str.replace(",","", regex=False))     # 그래서 다시 str 불러와 replace
)
# prod = prod[boolean]  : 조건(boolean)에 True 행 추출. 모든 컬럼 추출. => DataFrame 나옴.
prod = prod[prod["price"]>0]
```

```python
prod.info()
```

```python
# 마진, 마진비율 컬럼 가지고 있는 데이터 프레임 생성 : assign -> 컬럼 추가한 DataFrame 반환
prod2 = prod.assign(
    margin=prod["price"] - prod["cost"],
    margin_rate=(prod["price"] - prod["cost"]) / prod["price"]
)
```
